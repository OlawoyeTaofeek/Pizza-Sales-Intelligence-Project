```python
import pandas as pd
import os

# === Combine orders ===
order_files = ["orders_2023.csv", "orders_2024.csv", "orders_2025.csv"]
orders = pd.concat([pd.read_csv(f) for f in order_files], ignore_index=True)

# === Combine order_details ===
order_details_files = ["order_details_2023.csv", "order_details_2024.csv", "order_details_2025.csv"]
order_details = pd.concat([pd.read_csv(f) for f in order_details_files], ignore_index=True)

# === Load pizza data ===
pizzas = pd.read_csv("pizzas.csv")
pizza_types = pd.read_csv("pizza_type.csv")

# === Merge all together ===
merged_df = order_details \
    .merge(orders, on="order_id", how="left") \
    .merge(pizzas, on="pizza_id", how="left") \
    .merge(pizza_types, on="pizza_type_id", how="left")
    
merged_df.drop(columns=["order_details_id"], inplace=True)

# === Save full combined dataset ===
merged_df.to_csv("full_pizza_orders_flat.csv", index=False)

print(f"✅ Final merged dataset saved to: full_pizza_orders_flat.csv")
```
